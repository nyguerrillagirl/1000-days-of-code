```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm, uniform, expon, gamma, beta
```

```python
x = np.linspace(-4, 4, 400)
x_pos = np.linspace(0, 8, 400)
x_01 = np.linspace(0, 1, 400)
```

```python
plt.figure(figsize=(14, 10))

# Normal
plt.subplot(2, 3, 1)
plt.plot(x, norm.pdf(x))
plt.title("Normal Distribution")
```

```python
# Uniform
plt.subplot(2, 3, 2)
plt.plot(x_01, uniform.pdf(x_01))
plt.title("Uniform Distribution")
```

```python
# Exponential
plt.subplot(2, 3, 3)
plt.plot(x_pos, expon.pdf(x_pos))
plt.title("Exponential Distribution")
```

```python
# Gamma
plt.subplot(2, 3, 4)
plt.plot(x_pos, gamma.pdf(x_pos, a=2))
plt.title("Gamma Distribution (shape=2)")
```

```python
# Beta
plt.subplot(2, 3, 5)
plt.plot(x_01, beta.pdf(x_01, a=2, b=5))
plt.title("Beta Distribution (a=2, b=5)")

plt.tight_layout()
plt.show()
```
