```python
import numpy as np
from scipy.stats import binom, poisson
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
k = np.arange(0, 20)

plt.figure(figsize=(12, 5))

# Binomial
plt.subplot(1, 2, 1)
plt.stem(k, binom.pmf(k, n=20, p=0.4))
plt.title("Binomial Distribution (n=20, p=0.4)")

# Poisson
plt.subplot(1, 2, 2)
plt.stem(k, poisson.pmf(k, mu=5))
plt.title("Poisson Distribution (λ=5)")

plt.tight_layout()
plt.show()
```
